```python
# Using for loop to read the data from the url for each team abbrev
# Data2 has all the 3 related to 3 position , but it is a dict containing 3 diff list 
# then creating a single list called players, combining 3 lists into one and appending all the data into players_data 
# for some cases data is not present for few columns, so using 'get' helps us replace it with null and all the data is inserted.
players_data = []

for team_abbrev, team_id in team_lookup.items():
    url2 = f"https://api-web.nhle.com/v1/roster/{team_abbrev}/current"

    response2 = requests.get(url2)
    data2 = response2.json()

    players = data2['forwards'] + data2['defensemen'] + data2['goalies']

    for player in players:
        players_data.append(dict(
            player_id = player['id'],
            team_id = team_id,
            first_name = player['firstName']['default'],
            last_name = player['lastName']['default'],
            position = player['positionCode'],
            jersey_number = player.get('sweaterNumber'),
            birth_date =player.get('birthDate'),
            birth_country = player.get('birthCountry'),
            height_cm = player.get('heightInCentimeters'),
            weight_kg = player.get('weightInKilograms'),
            shoots_catches = player.get('shootsCatches'),
            headshot_url = player.get('headshot')))
    
```
